Problem 1: Mobile Recharge Validation System
 Problem Statement
Create a function to validate a mobile recharge request.
Rules:
Recharge amount must be ≥ ₹50


Valid plans: 199, 299, 399, 599


User should be allowed to retry if invalid


Real-Time Use
Telecom recharge apps


Payment gateways
*italicised text*

```python
# Function to validate recharge amount and plan
def validate_recharge(amount, plan):
    valid_plans = [199, 299, 399, 599]

    if amount < 50:
        print(" Recharge amount must be at least ₹50")
        return (False)

    if plan not in valid_plans:
        print(" Invalid plan selected")
        return False

    print("Recharge Successful!")
    return True


# Retry using while loop
while True:
    amount = int(input("Enter recharge amount: ₹"))
    plan = int(input("Enter plan (199, 299, 399, 599): "))

    if validate_recharge(amount, plan):
        break   # stop loop if valid
    else:
        print("Please try again...\n")
```

Problem 2: Inventory Reorder Alert System
Problem Statement
Given a dictionary of products and stock quantities:
Print Reorder Alert if stock < 15


Print Stock OK otherwise


Real-Time Use
Warehouse systems


Retail management


```python
# Function to check stock levels
def check_inventory(products):
    for item, quantity in products.items():
        if quantity < 15:
            print(item, "→ Reorder Alert ")
        else:
            print(item, "→ Stock OK ")


# Sample dictionary
inventory = {
    "Rice": 20,
    "Sugar": 10,
    "Oil": 5,
    "Soap": 18
}

check_inventory(inventory)
```

 Problem 3: Student Result Processing System
Problem Statement
Create a function that:
Accepts a list of student marks


Calculates average marks


Prints Pass if average ≥ 50 else Fail


Real-Time Use
Online exam systems


Evaluation portals


```python
# Function to calculate result
def check_result(marks):
    total = 0

    # Calculate total using loop
    for m in marks:
        total += m

    average = total / len(marks)
    print("Average Marks:", average)

    if average >= 50:
        return ("Pass ")
    else:
        return ("Fail")


# Example
student_marks = [60, 55, 40, 70, 50]
result = check_result(student_marks)
print("Result:", result)
```

Problem 4: Cab Fare Estimator with Retry Option
Problem Statement
Create a cab fare calculator:
Base fare = ₹50


₹12 per km


Peak hour → 25% extra


Allow retry using while loop


Real-Time Use
Ride-hailing apps


Transport systems




```python
# Function to calculate fare
def calculate_fare(distance, peak):
    base_fare = 50
    per_km = 12

    fare = base_fare + (distance * per_km)

    if peak.lower() == "yes":
        fare = fare + (fare * 0.25)  # 25% extra

    return fare


# Retry option
while True:
    distance = float(input("Enter distance in km: "))
    peak = input("Is it peak hour? (yes/no): ")

    total_fare = calculate_fare(distance, peak)
    print("Total Fare: ₹", total_fare)

    retry = input("Do you want to calculate again? (yes/no): ")
    if retry.lower() != "yes":
        break
```

Problem 5: Employee Attendance Eligibility Checker
Problem Statement
Create a function that:
Accepts attendance list ("P", "A")


Calculates attendance percentage


Returns Eligible if ≥ 75% else Not Eligible


Real-Time Use
HR payroll systems


Internship tracking




```python
# Function to check attendance eligibility
def check_attendance(attendance_list):
    present_days = 0

    # Count present days
    for day in attendance_list:
        if day == "P":
            present_days += 1

    percentage = (present_days / len(attendance_list)) * 100
    print("Attendance Percentage:", percentage, "%")

    if percentage >= 75:
        return ("Eligible ")
    else:
        return ("Not Eligible ")


# Example
attendance = ["P", "P", "A", "P", "P", "A", "P", "P"]
status = check_attendance(attendance)
print("Status:", status)
```

Problem 6: Password Strength Checker
Problem Statement
Create a function to check password strength.
Rules:
Minimum length = 8


Must contain at least one digit


Must contain at least one special character (@#$)


Real-Time Use
Login systems


Security validation


```python
# Function to check password strength
def check_password(password):
    has_digit = False
    has_special = False
    special_chars = "@#$"

    if len(password) < 8:
        return (" Password must be at least 8 characters")

    # Loop through password characters
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in special_chars:
            has_special = True

    if not has_digit:
        return ("Password must contain at least one digit")

    if not has_special:
        return ("Password must contain at least one special character (@#$)")

    return ("Strong Password")


# Example
pwd = input("Enter password: ")
print(check_password(pwd))
```
