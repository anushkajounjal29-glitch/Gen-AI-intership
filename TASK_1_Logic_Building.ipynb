2. Pass / Fail Analyzer
Problem Statement:
Analyze student results.
marks = [45, 78, 90, 33, 60]

Requirements:
A student passes if marks ≥ 50
Count the total number of pass students
Count the total number of fail students
Print the final result clearly
Real-World Application: Academic evaluation systems


```python
#marks from user
marks=[45,78,90,33,60]

pass_count=0
fail_count=0
#loop for counting
for mark in marks:
    print("checking",mark)

    if mark>=50:
        pass_count+=1
    else:
        fail_count+=1
#rint result once
print("\npass/fail analyzer")
print("total pass students:",pass_count)
print("total fail students:",fail_count)
#final class performamce
if pass_count>fail_count:
    print("final result: class performance is Good")
elif pass_count==fail_count:
    print("final result:class performance is average")
else:
    print("final result:class performance needs improvement")
```

3. Simple Data Cleaner
Problem Statement:
Clean and standardize user names.
names = [" Alice ", "bob", " CHARLIE "]

Requirements:
Remove extra spaces
Convert all names to lowercase
Print the cleaned list
Real-World Application: Data preprocessing before analysis



```python
#raw data with conssistency
names=["alice","boB","CHARLI"]
cleaned_names=[]#new list store
#cleaning each name
for name in names:
    cleaned_name=name.strip().lower()
    #standarize case
    cleaned_names.append(cleaned_name)
   #print cleaned data
    print("cleaned names list")
    print(cleaned_names)
```

4. Message Length Analyzer
Problem Statement:
Analyze message sizes.
messages = ["Hi", "Welcome to the platform", "OK"]

Requirements:
Print the length of each message
Flag messages longer than 10 characters
Real-World Application: Text filtering and validation systems


```python
#chexk message length
message =["hi","welcome to the platform","ok"]
print("message length analysis")
#for each message
for message in message:
    length=len(message)#length calculate no of character
    print(f"message:'{message}' |length'{length}'")
#flag message longer than 10 character
    if length>10:
      print("->this message longer than 10 charaxters.")
```
